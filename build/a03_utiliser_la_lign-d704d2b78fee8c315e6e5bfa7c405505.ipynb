# Utiliser la ligne de commandes dans un Terminal

La ligne de commande dans un terminal est souvent perçue comme quelque chose de barbare, compliqué et inutile car on peut souvent faire la même chose avec de belles boîtes de dialogues, des boutons et zones de saisies dans une interface graphique... Et c'est juste.


~~~{image} images/terminal/terminal_boss.png
:width: 500px
:align: center
~~~



## Pourquoi utiliser un terminal ?

Le terminal n'est pas réservé aux informaticiens aguerris ou pirates de tous bords; il vous sera utile dès que vous souhaitez reprendre le contrôle sur votre PC.

Les terminaux avec leurs commandes sont les ancêtres des explorateurs de fichiers.

Elles permettent d'exécuter des actions sur les fichiers de votre ordinateur:

* Création, suppression de dossier
* Création, modification, renommage, déplacement et suppression de fichiers
* Se déplacer dans les dossiers et sur les disques durs
* Rechercher des fichiers

Mais vous pouvez faire beaucoup plus de choses comme:

* Modifier la date système
* Exécuter des programmes
* Changer la configuration réseau 
* Vous connecter sur d'autres machines
* ...

Le terminal vous sera utile dès que:

* Vous programmez avec des langages comme C, Java, Python...
* Vous avez besoin de comprendre ce qui ne va pas sur votre PC
* Vous avez besoin de démarrer un programme avec des options/paramètres que ne permet pas de préciser l'explorateur de fichiers







## Dans un terminal c'est vous qui commandez !

> Un peu de prosélytisme... Ce qui suit n'est pas de la plus grande objectivité.

Sur les ordinateurs utilisant des systèmes d'exploitations propriétaires vous subissez très souvent le bon vouloir du fabricant/de l'éditeur:  

* L'ordinateur vous impose des mises à jours sans trop savoir ce qu'elles vont faire,
* Il vous impose les moments ou il redémarre
* Personne ne vous dit ce qui se passe en arrière plan avec vos informations personnelles...
* Trop de choses sont faîtes afin que vous restiez un simple utilisateur qui ne se pose pas de questions...

Bref, *vous subissez votre ordinateur plus que vous ne le contrôlez !*

Ce n'est pas normal, une machine doit être conçue pour nous obéir, elle doit nous faciliter la vie et non nous imposer ce qu'aura décidé un éditeur de logiciel. La machine est à vous, c'est votre matériel, c'est uniquement à vous qu'elle doit obéir ! 

**Avec le terminal, vous reprennez les commandes.**

Tous les systèmes d'exploitation ne sont pas ainsi, et ceux qui sont basés sur des logiciels libres, comme Linux/FreeBSD, sont conçus dans cette optique : l'ordinateur doit faire ce que vous lui demandez et vous avez le plein contrôle sur ce qui se passe. 

Aussi, avec Linux, l'usage du terminal est quasi quotidien ! Car tout peut se faire en ligne de commande. Vous pouvez tout contrôler. Et vous verrez, finalement c'est facile...

## Quelques termes à connaître et remarques

Avant de lancer nos premières commandes, expliquons quelques termes que nous utiliserons fréquemment.

~~~{admonition} Prompt
:class: note

Le **prompt** du terminal désigne la ligne courante sur laquelle le texte que vous saisissez s'affiche.  
Ce prompt est matérialisé par un texte qui est personnalisable et dont la valeur par défaut diffère selon votre système d'exploitation.

* `C:\Utilisateurs\User> ` sous Windows
* `user@localhost bash$ ` sous Linux
* `user@MacBook-Pro ~ % ` sous Mac
~~~

~~~{admonition} OS
:class: note

Le terme **OS** sera utilisé pour désigner votre système d'exploitation (Operating System), c'est à dire Windows, Mac ou Linux.
~~~


~~~{admonition} Sortie courante
:class: note

Le terme **sortie courante** désignera le texte affiché suite à l'exécution d'une commande
~~~

~~~{admonition} Console
:class: note

Le terme **console** désigne le terminal et/ou la zone de saisie.  
Souvent vous entendrez des phrases comme 

> saisissez dans la console Unix/MSDOS la commande suivante: ...

Cela signifie que vous devez démarrer un terminal et taper cette commande

~~~




~~~{warning} 
Nous utiliserons maintenant le prompt généric `shell/C:> ` pour désigner le *prompt* de la ligne courante dans votre terminal.  

Vous ne devrez pas saisir ce texte, mais celui qui le suit.
~~~

~~~{admonition} Exécution d'une commande
:class: tip
Pour exécuter une commande, il faut saisir son nom sur la ligne du prompt et appuyer sur `<ENTREE/RETURN>`  

Les commandes s'écrivent en général sur une seule ligne, le texte affiché sur les lignes qui suivent sont le résultat de ces commandes
~~~

## Premières commandes - Manipuler les fichiers

Nous allons pouvoir exécuter les premières commandes dans notre terminal. Commençons par apprendre à nous déplacer sur le disque dur et à manipuler les fichiers et dossiers.

### Connaître son emplacement

Quand un terminal démarre, il se place dans un dossier du disque dur, en général ce dossier est votre dossier personnel.

Pour connaître ce dossier, saisir la commande suivante dans le terminal:

~~~~{admonition} Windows
:class: note

Avec `CMD`

~~~bash
shell/C:> echo %cd%
C:\Users\Utilisateur
~~~

Avec `PowerShell`

~~~bash
shell/C:> pwd
Path
----
C:\Users\Utilisateur
~~~


~~~~

~~~~{admonition} Linux ou Mac
:class: note

Ces 2 systèmes sont des systèmes de type [Unix](https://fr.wikipedia.org/wiki/Unix), ils partagent les mêmes commandes:

~~~bash
shell/C:> pwd
/home/user
~~~


~~~~

### Changer de dossier

Dans tous les terminaux la commande est la même : `cd <nom du dossier>`.

Le nom du dossier sera spécifique à votre OS.  S'il contient des espaces ou des caractères spéciaux, il faudra l'entourer de guillemets.

En général, changer de dossier modifie le prompt qui affiche ce nouvel emplacement.

~~~~{admonition} Windows
:class: note

~~~bash
shell/C:> cd c:\windows
c:\Windows> <nouvelle commande ici>
~~~

~~~~

~~~~{admonition} Unix (Linux ou Mac)
:class: note

~~~bash
shell/C:> cd /var/log
shell/C: /var:log> <nouvelle commande ici>
~~~

~~~~

Quand vous ne connaissez pas le dossier ou vous voulez vous rendre, vous pouvez en général le retrouver avec l'explorateur de fichiers et le copier/coller dans le terminal:

Ou Unix:

~~~{warning}

N'oubliez-pas : Si le chemin de votre dossier contient des caractères spéciaux comme des espaces, il faut l'entourer de guillemets:

* `cd "c:\dossier 1\dossier 2\mon dossier a moi"` 
* `cd "/dossier 1/dossier 2/mon dossier a moi"` 
~~~

~~~{note}

Sous Windows le séparateur des chemins dans les noms de dossiers est le caractère `\`, sous Unix c'est `/`.

Vous pouvez aussi utiliser `/`  sous Windows.

### Pourquoi lancer un programme depuis le terminal ?

Lancer un programme depuis le terminal permet plusieurs choses:

* Avoir un message d'erreur quand cela se passe mal  
  En effet, avec l'explorateur de fichiers souvent quand le programme refuse de démarrer, rien ne s'affiche à l'écran.  
  En lançant le programme avec le terminal vous aurez plus de chances d'obtenir un message d'erreur dans la console
  
* Passer des options au programme que vous ne pouvez pas préciser en cliquant dessus depuis un explorateur de fichiers

* Ecrire une série de commandes dans un script pour automatiser des tâches répétitives

* ...

### Modifier/Créer une variable d'environnement

Il est parfois utile de modifier une variable d'environnement existante ou de créer les siennes.

La variable d'environnement la plus utile est probablement `PATH` : elle contient la liste des dossiers ou votre terminal ira chercher les programmes à exécuter.

Si votre programme se trouve dans un de ces dossiers, vous n'avez pas besoin d'écrire son chemin complet pour le démarrer, vous pouvez juste écrire son nom, et même sans le `.exe` sous windows 

~~~bash
Shell/C:> notepad
~~~

Pour consulter la valeur d'une variable seule, utilisez les commandes suivantes.

#### Sous Windows

Avec `CMD` :  `echo %NOM_VARIABLE%`

## Modifier 

Pour la modification, il convient d'utiliser une autre commande encore.
Deux cas de figures peuvent se présenter:


1. Soit vous souhaitez affecter une toute nouvelle valeur à la variable
2. Soit vous souhaitez compléter la valeur existante.  
   Dans ce cas vous réutilisez le nom de la variable d'environnement dans la valeur affectée

Pour la variable PATH par exemple, vous souhaitez en général la compléter avec l'emplacement d'un nouveau programme que vous venez d'installer, comme *python* ou *java*




### Sous Unix:

~~~bash
export PYHTONPATH=$PYTHONPATH:/autre/chemin1:/autre/chemin2
~~~

### Sous Windows:

Avec `CMD.exe`
~~~bash
set PYHTONPATH=%PYTHONPATH%;c:\autre\chemin1;c:\autre\chemin2
~~~

Avec `Powershell`
~~~bash
$env:PYTHONPATH=$env:Path;c:\autre\chemin1;c:\autre\chemin2
~~~


## Lister l'emplacement d'un programme

Si vous avez plusieurs version d'un programme installé il est intéressant de savoir lequel est utilisé.

### Sous Unix

~~~bash
bash$ which python
/usr/bin/python
~~~

### Sous Windows

Avec `CMD.exe`
~~~cmd
c:\> where python
c:\python3\python.exe
c:\users\login\miniforge3\python.exe
~~~

Le premier affiché est celui utilisé.

Avec `Powershell`
~~~bash
Get-Command: python
~~~
