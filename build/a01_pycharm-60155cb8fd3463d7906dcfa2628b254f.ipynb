Pour suivre ce cours vous pouvez aussi utiliser l'éditeur [PyCharm](https://www.jetbrains.com/pycharm/) en version communautaire.


PyCharm est un Environnement de Développement Intégré dédié à Python.

Il permet d'écrire un programme en Python, de l'exécuter, le déboguer et propose toutes les fonctionnalités de tels outils:

* Coloration syntaxique
* Complétion de code
* Aide et recherche
* Versionnement du code source
* ...




## Autres IDE
Il existe beaucoup d'autres IDE pour Python.

* [VSCode](https://code.visualstudio.com/) de Microsoft  
  Cet éditeur propose l'avantage de pouvoir écrire vos programmes dans de multiples langages et fonctionne sous Mac, Linux et Windows.  Il est aussi Open Source.
  
  Notre [Annexe Visual Studio Code](a02_visual_studio_code.ipynb) présente comment l'installer et le configurer.
  
* [SublimeText](https://www.sublimetext.com/)  très apprécié des personnes utilisant un Mac.  
  Il est léger et efficace mais n'offre pas toutes les fonctionnalités d'un IDE comme le déboguage.
  
* Cette page du Wiki Python en présente [beaucoup d'autres](https://wiki.python.org/moin/PythonEditors)

### Démarrez l'installation

![Install PyCharm 1](images/pycharm/pycharm_windows_1.png)

Acceptez le dossier d’installation par défaut ou choisissez en un autre si vous avez des préférences.

Cochez les option « Create desktop shortcut » et « Create association »

Puis cliquez sur « next » et « install »

![Install PyCharm 2](images/pycharm/pycharm_windows_2.png)



## Lancez Pycharm

Pour démarrer PyCharm, procédez comme suit:

* Sous Windows  
  Utilisez l’icône du bureau
* Sous Unix  
  Utilisez le raccourci du bureau ou saisir la commande `pycharm.sh` en précisant son chemin d'installation: `/dossier/de/pycharm/bin/pycharm.sh` depuis un terminal.
  
 
Pour une première utilisation, ne pas importer la configuration


![Install PyCharm 3](images/pycharm/pycharm_windows_3.png)

## Création de votre projet

Un projet est un dossier dans lequel vous écrirez vos programmes.  
Afin d'utiliser PyCharm il convient de lui indiquer:

* Dans quel dossier travailler
* Avec quelle version de Python, vous pourriez avoir plusieurs Python sur votre PC.


![Install PyCharm 4](images/pycharm/pycharm_windows_4.png)

Cliquez sur *New Project*

Sélectionnez le dossier dans lequel vous souhaitez travailler puis l’interpréteur Python Anaconda3.

Pour cela :

* Cochez « previously configured interpreter »
* Puis sélectionnez 
    * le programme `python.exe` du dossier d'installation d’Anaconda sous Windows 
    * ou `/chemin/vers/anaconda/bin/python` sous Unix

![Install PyCharm 5](images/pycharm/pycharm_windows_5.png)

Si vous ne trouvez pas votre *Python* dans la liste déroulante cliquez sur les 3 « ... » et sélectionnez l’emplacement de votre exécutable python :

![Install PyCharm 6](images/pycharm/pycharm_windows_6.png)

Si vous ne vous rappelez plus de l'emplacement d'Anaconda, vous pouvez le retrouver en lançant un terminal sous Unix ou une console *Anaconda Prompt* sous Windows. Ensuite dans ce terminal saisir les commandes ci-après:

```bash
bash$ ou C:> python
Python 3.8.8 (default, Apr 13 2021, 19:58:26) 
[GCC 7.3.0] :: Anaconda, Inc. on linux
Type "help", "copyright", "credits" or "license" for more information.
>>> import sys
>>> sys.executable
'/home/<user>/anaconda3/bin/python'
```

L'emplacement affiché est celui du programme Python qu'il faut sélectionner.

Cliquez sur « ok » puis « create »

## Si vous avez oublié de sélectionner votre Python

Si vous n’avez pas pu sélectionner votre exécutable *Python* lors de la création de votre projet, PyCharm en aura installé un nouveau spécialement pour votre projet dans un dossier « venv ».

Supprimez-le avec l’explorateur de fichiers et retournez dans la configuration de Pycharm pour le changer :

* Sous Linux et Windows :
  Menu File puis Settings
* Sous Mac
  Menu Préférences puis Settings
  
  
![Install PyCharm 8](images/pycharm/pycharm_windows_8.png)

Allez ensuite dans le menu « Project/Python interpreter »

![Install PyCharm 9](images/pycharm/pycharm_windows_9.png)

Cliquez sur l’engrenage à droite de votre interpréteur *python*, sélectionnez « add » puis recommencez la procédure de sélection de l’exécutable Python installé par Anaconda.

![Install PyCharm 10](images/pycharm/pycharm_windows_10.png)
