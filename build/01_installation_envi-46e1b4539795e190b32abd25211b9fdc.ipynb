## Miniforge

[Miniforge](https://github.com/conda-forge/miniforge) est une version allégée d'Anaconda, incluant un python minimaliste et les commandes de base d'Anaconda.

Il est spécifique à conda-forge. Miniforge est configuré avec les fonctionnalités suivantes :

* *conda-forge* est le seul canal de téléchargement
* Support des interpréteurs [Python](https://www.python.org/) (C-Python) et [Pypy](https://www.pypy.org/)
* Support optionnel de Mamba à la place de Conda.
* Support de différentes architectures CPU (x86_64, ppc64le, et aarch64 y compris Apple M1).


## Installation de Miniforge

### Sous Linux et Mac

Nous vous conseillons l'installation en ligne de commandes en utilisant le fichier `.sh` que vous pouvez télécharger sur la forge  [Miniforge](https://github.com/conda-forge/miniforge)

Choisissez-bien la version `python` **et NON** `pypy`



Ouvrez un terminal et exécutez le script d'installation:

```bash 
bash$ bash <script mamba/miniforge>.sh
```

`<script mamba/miniforge>` doit être remplacé par le nom du script téléchargé, par exemple `Miniforge3-Linux-x86_64.sh`.

Nous vous invitons à accepter les choix par défaut:


* Choisissez un dossier d’installation facile dans lequel vous avez les droits comme `$HOME/miniforge3` 
* Activez le shell Mamba/Miniforge.  
  Si vous ne l’avez pas fait au moment de l'installation, dans un terminal saisir:   
  `bash$ $HOME/miniforge3/bin/mamba init <votre shell comme bash pour linux ou csh sous mac>`
* Relancez le terminal, le prompt de votre shell doit contenir le mot `(base)` 

Vérifiez avec la commande `which` (`where` sous Windows) l'emplacement de python, puis démarrez python et saisir les commandes ci-après:

~~~
(base) shell$ which python
/home/user/miniforge/bin/python
(base) shell$ python
Python 3.12.5 | packaged by conda-forge | (main, Feb  2 2023, 20:20:04) [GCC 11.3.0] on linux
Type "help", "copyright", "credits" or "license" for more information.
>>> import sys
>>> sys.executable
'/home/user/miniforge/bin/python'
~~~

Ici la version de python est la 3.12.5, il se peut que vous en ayez une plus récente.

### Sous Windows

Sous Windows,

[Téléchargez l'exécutable 64 bits depuis la forge](https://github.com/conda-forge/miniforge#miniforge3) puis lancez l'application `Miniforge3-Windows-x86_64.exe`

Suivez ensuite les instructions ci-après pour l'installation:

```{image} images/mamba/miniforge-windows-1.png
:width: 500px
:align: center
```


Dans le prompt, saisir `python`  puis les commandes suivantes, une fois l'interpréteur démarré:

~~~python
>>> import sys
>>> sys.executable
'C:\\Users\\Utilisateur\\miniforge3\\python.exe'
~~~

Ceci affichera l'emplacement de l'exécutable Python, vous pourriez en avoir besoin par la suite.

## Installer un environnement virtuel Python

Un environnement virtuel Python consiste à installer une version de Python que l'on destine à un projet particulier:

* Cela n'a rien à voir avec la virtualisation est des outils comme VMWare, VirtualBox, Docker...
* Il s'agit s'implement d'installer un nouveau python dans un nouveau dossier
* L'idéal étant d'utiliser un python différent par projet
* Ainsi chaque projet peut faire évoluer sa version de python et des librairies dont il a besoin sans imposer ses choix aux autres projets qui risqueraient alors de ne plus fonctionner avec cette nouvelle configuration.


### Pourquoi un environnement virtuel ?

~~~{admonition} Un Python par projet
:class: tip
**Chaque projet doit avoir son propre Python.**

Afin de ne pas créer de conflits entre vos projets lors de la mise à jour d'une librairie ou de l'interpréteur Python pour un projet donné, que vous ne souhaiteriez pas faire évoluer de la même manière pour vos autres projets, il devient incontournable d'utiliser des [environnements virtuels](https://packaging.python.org/en/latest/tutorials/installing-packages/#creating-and-using-virtual-environments)
~~~  
  

### Les outils pour créer un environnement virtuel

Pour créer des environnements virtuels plusieurs outils existent:

* [virtualenv](https://packaging.python.org/key_projects/#virtualenv)
* [venv](https://docs.python.org/3/library/venv.html)
* [pipenv](https://packaging.python.org/key_projects/#pipenv)
* [conda](https://packaging.python.org/key_projects/#conda)/[mamba](https://mamba.readthedocs.io/en/latest/)
* [pipx](https://pipx.pypa.io)
* [poetry](https://python-poetry.org/)
* [uv](https://docs.astral.sh/uv/)
* [pixi](https://pixi.prefix.dev)

Avec Mambaforge la commande officielle est `mamba`.

Avec Miniforge la commande officielle est `conda`.

Dans les exemples ci-après vous pouvez intervertir les commandes `mamba/conda` selon votre choix de distribution.

Les commandes `conda` et `mamba` savent à la fois installer des environnements virtuels, mais aussi des librairies. Elles sont un mélange des outils *pip* et *virtualenv* en un peu plus intelligent et polyvalent car elles savent aussi installer d'autres logiciels que Python et ss librairies.

Pour une bonne compréhension nous vous invitons à la lecture de cet article [Conda: Myths and Misconceptions](https://jakevdp.github.io/blog/2016/08/25/conda-myths-and-misconceptions/)

### Créer un environnement virtuel

Nous allons donc créer un environnement virtuel pour notre formation, ceci évitera de modifier celui installé par défaut par Miniforge et de tout casser si une nouvelle librairie venait à créer des conflits avec celles déjà installées. Seul notre environnement serait endommagé et Miniforge continuerait de fonctionner normalement.

~~~{tip}
Pour la suite des commandes, privilégiez la commande `mamba` à `conda` pour plus de rapidité.  

*Mais vous pouvez intervertir sans conséquence ces 2 commandes à votre gré*

Dans quelques rares cas la commande `mamba` ne possède pas encore les options équivalentes de `conda`. Dans ce cas, revenez sur `conda`. 
~~~

Pour créer un environnement virtuel la commande à utiliser est :
```
mamba create --name <nom_environnement> python=<version> <éventuellement des librairies>
```

Vous pouvez ajouter à cette commande le paramètre `-c conda-forge` qui permet de télécharger les librairies sur la forge communautaire plutôt que sur les serveurs principaux (main) d'Anaconda. Cela présente plusieurs avantages:

* Par défaut les serveurs *pkg-main* d'Anaconda ne proposent qu'environ 2000 librairies, alors que les serveurs *conda-forge* en proposent bien plus
* Les versions des librairies sur *conda-forge* sont généralement bien plus récentes
* La forge communautaire n'est pas payante

Toutefois avec *mamba* et *miniforge* ce choix est celui par défaut, l'option n'est donc requise que si vous utilisez encore Anaconda.

**Ouvrez un terminal Unix/Miniforge prompt et saisir cette commande:**

Nous utiliserons l'avant dernière version de python pour éviter les conflits: fréquemment les nouveautés de la dernière version de Python empêchent certaines librairies scientifiques de fonctionner. 

```bash
mamba create --name "formation" -c conda-forge  python=3.14
```

~~~{admonition} Attention avant de valider la commande
:class: warning
Avant de valider la commande, vérifiez tout de même que le nom de la forge proposée est bien *conda-forge* et non *pkg-main*, sinon ajoutez l'option `-c conda-forge` 
~~~

### Installation des librairies

Pour installer les librairies nous allons procéder en 4 étapes.

1. Activer l'environnement virtuel pour pouvoir utiliser sa version de Python
2. Optionnel: Définir les serveurs de `conda-forge` comme choix par défaut  
3. Installer les librairies
4. Ajouter l'environnement dans le menu de Jupyter

#### Activation de l'environnement virtuel

Vous l'aurez peut-être remarqué, mais si vous saisissez `python` dans le terminal juste après la création de l'environnement virtuel, c'est toujours le python par défaut de `Miniforge/Mambaforge` qui démarre.

Vous pouvez le vérifier en saisissant ces commandes:

~~~python
>>> import sys
>>> sys.executable
'C:\\Users\\vous\\miniforge3\\python.exe'
~~~

Le fait d'installer un nouveau python ne le rend pas disponible dans le terminal, il faut pour cela l'activer avec la commande:   
`mamba activate <nom du python>` 

Pour lister les environnements déjà installés, saisir la commande suivante (après avoir quitté l'interpréteur) dans le prompt de votre terminal:

~~~bash
shell ou c:> mamba env list
# conda environments:
#
base                  *  /home/user/mambaforge
formation            /home/user/mambaforge/envs/formation
~~~

`shell ou c:>` représente le prompt de votre shell, vous ne devez pas le saisir...


```bash
shell ou c:> mamba activate formation
(formation) shell ou c:>
```

Maintenant, si vous ressaisissez la ligne `sys.executable` vous observerez que vous utilisez la bonne version de python:

~~~bash
(formation) shell ou C:> python
Python 3.14.6 | packaged by conda-forge | (main, Jul  2 2026, 20:20:04) [GCC 11.3.0] on Windows
Type "help", "copyright", "credits" or "license" for more information.
>>> import sys
>>> sys.executable
'C:\\Users\\vous\\miniforge3\\envs\\formation\\python.exe'
~~~

```bash
(formation) conda config --env --add channels conda-forge
```

## Note pour les utilisateurs de Windows

Il arrive souvent sous Windows que `jupyter` n'arrive pas à charger les librairies/dll d'un environnement virtuel quand il est lancé depuis l'environnement de base.

Pour contourner ce problème il est possible d'installer jupyter dans cet environnement virtuel et de le démarrer depuis ce dernier.

```bash
(formation) mamba install jupyter jupyterlab  # mais nous l'avons déjà fait ci-dessus
```

## Ajouter l'environnement au menu Jupyter
Enfin, afin de disposer de ce nouvel environnement dans les menus de Jupyter/Jupyterlab, il convient de l'y ajouter.

**Depuis l'environnement virtuel activé**, saisir ces commandes:

```bash
(formation) mamba install  ipython ipykernel  # Normalement déjà fait
```

Puis:

```bash
(formation) python -m ipykernel install --user --name=<nom environnement>  --display-name="<nom affiché dans Jupyter>"`
```

`<nom environnement>` représente le dossier qui sera créé dans la configuration de `ipykernel`, le nom de votre environnement sera un bon choix.

Dans notre cas, la commande finale sera:

```bash
(formation)  python -m ipykernel install --user --name=formation --display-name="Formation (Python 3.14)"
```

## Supprimer un environnement

Vous souhaitez supprimer un environnement qui n'est plus utilisé ou recommencer votre installation, vous pouvez le faire avec cette commande

~~~bash
(base) mamba remove --name <environnement> --all
~~~

Exemple:
~~~bash
(base) mamba remove --name formation --all
~~~
