[Visual Studio Code](https://code.visualstudio.com/), souvent abrégé en VS code, est un éditeur de code édité par Microsoft, fonctionnant sous Linux, Windows et MacOS. C'est un éditeur très complet, comprenant de nombreuses fonctionnalités comme le support de git dans l'éditeur de code, la complétion automatique du code, ou encore la coloration syntaxique, un débugger, une fonctionnalité de refactoring du code. Nous verrons aujourd'hui comment le configurer sous Linux et le paramétrer au mieux pour l'utiliser avec Python. 

## Installation

Commencez par [télécharger le logiciel depuis le site vscode](https://code.visualstudio.com/#alt-downloads)
puis suivez [la procédure d'installation de votre plateforme](https://code.visualstudio.com/docs/setup/setup-overview)

Pour des questions de confidentialité et sécurité, nous vous conseillons d'installer la version [VSCodium](https://vscodium.com/)

Si vous ne savez plus/pas ou se trouve l'exécutable de votre Python, dans un terminal avec l'environnement activé, lancez l'interpréteur Python puis saisir ces commandes:

```python
bash ou c:> python
>>> import sys
>>> sys.executable
<affiche l'emplacement de votre python>
```

### Installation d'un linter

**Qu'est-ce qu'un linter ?**

Un linter est un outil d'analyse statique du code, permettant de mettre en évidence des problèmes syntaxiques et stylistiques, ce qui vous permet d'identifier et de corriger des erreurs de programmations subtiles, ou des formulations qui ne sont pas conventionnelles et peuvent amener, plus tard, à des erreurs ou des bugs. 

Un linter est différent d'un formateur, le linter analyse la façon dont le code s'exécuterait et indique les corrections en fonction, là où les outils de formattage s'attachent seulement à l'apparence du code. 

Aujourd'hui, presque tout les projets utilisent un linter, ce qui permet d'améliorer la qualité du code et la compatibilité des différentes version d'un projet. Pour tous les langages de programmation, il existe plusieurs linters. 

Pour Python, les linters supportés par VScode sont [pylint](https://pypi.org/project/pylint/), [Flake8](https://flake8.pycqa.org/en/latest/), [Pylama](https://pypi.org/project/pylama/1.4.0/), [mypy](http://mypy-lang.org/), [Bandit](https://bandit.readthedocs.io/en/latest/), [pycodestyle](https://pypi.org/project/pycodestyle/), et [Prospector](https://prospector.landscape.io/en/master/index.html).

Dans ce tutoriel, nous allons utiliser Pylint.  

**1. Autoriser le linter**

Ouvrez la fenêtre de commandes avec `ctrl + shift + p`, et sélectionner `Python: Select Linter`. Choisissez un des linters dans la liste proposée (ici, Pylint).

![10.png](images/vscode/10.png)

**2. Installer le linter**

Le linter est autorisé, mais pas installé. Lorsque vous sélectionnez un linter, VScode vous ouvre une fenêtre pour vous informer de ce fait. **Cliquez sur "Installer"**.

![11.png](images/vscode/11.png)

Le linter est maintenant installé ! **Pour le désinstaller**, il vous suffira de retourner dans les paramètres avec `ctrl + shift + p`, et sélectionner "Python:Enable/Disable Linting". 

### Utilisation du linter

Pour utiliser le linter, copiez collez le code ci-dessous dans votre programme (ou utilisez un autre programme de votre choix), puis utilisez la palette de commandes (`ctrl + shift + P`), cherchez "linting", et sélectionnez **"Python: Run Linting"**. 

Voici un exemple de programme à votre disposition pour tester le linter : 
```python
from math import sqrt
import time
def nombres_premiers(nb):
    i = 2
    liste = list()
    while nb % 2 == 0:
        liste.append(2)


        nb = nb/2
    for i in       range(3, int(sqrt(nb))+1, 2):
        while nb % i == 0:
            liste.append(i)
            nb = nb/i
    if nb>2:
        liste.append(nb)
    return (liste)
    



nombres_premiers(13195)
nombres_premiers(600851475143)
    
```
Ce programme permet de retrouver les facteurs premiers d'un nombre `nb`. Ses défauts comportent :
* Une docstring absente : on ne sait pas ce que fait ce code à moins de passer du temps à le lire et essayer de le comprendre. 
* Un print au lieu d'un return : la fonction pourrait retourner la liste, mais l'affiche à la place. 
* Le nombre de lignes blanches avant et après la fonction n'est pas conforme à la PEP8 (deux lignes avant et deux lignes après doivent rester blanches
* Le module time ne sert à rien, il n'est pas utilisé, il devrait être supprimé. 

Voici le résultat en utilisant deux linters différents : 

1. Avec Pylint

![12.png](images/vscode/12.png)

2. Avec Pylama

![13.png](images/vscode/13.png)

Comme vous pouvez le voir sur ces images, les deux linters remarquent des choses différentes : en effet, Pylama est en réalité un paquet comprenant plusieurs outils : il possède un linter, et plusieurs formateurs de code suivant plusieurs normes différentes (PEP8, PEP257). Pylama utilise Pylint mais aussi PyFlakes. Il est donc plus complet. 

Voici la fonction telle qu'elle est après toutes les corrections indiquées par Pylama : 

```python
from math import sqrt


def nombres_premiers(nb):
    i = 2
    liste = list()
    while nb % 2 == 0:
        liste.append(2)
        nb = nb / 2
    for i in range(3, int(sqrt(nb)) + 1, 2):
        while nb % i == 0:
            liste.append(i)
            nb = nb / i
    if nb > 2:
        liste.append(nb)
    return (liste)


nombres_premiers(13195)
nombres_premiers(600851475143)
```

Ce n'est pas parfait, il manque encore une docstring, et le nom de la fonction n'est pas très informatif quant à son utilité, elle devrait plutôt s'appeler "facteurs_premiers". Mais le code reste beaucoup plus propre qu'au début ! Le linter facilite les revues de code.
