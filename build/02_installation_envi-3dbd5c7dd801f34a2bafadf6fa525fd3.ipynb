## Pixi

Nous vous proposons d'installer votre environnement de travail avec [Pixi](https://pixi.prefix.dev).


*Pixi* est un gestionnaire de paquets, permettant de gérer des environnements de développements pour tout langage : Python, R, Rust, Javascript...

Il est rapide et respecte les bonnes pratiques de chaque langage.

Pixi va nous servir à installer Jupyter et les autres librairies du projet.


## Deno

Le plugin JavasSript pour jupyter, appelé *kernel javascript* que nous vous proposons est [Deno](https://deno.com/).  
Il en existe d'autres.

~~~{figure} https://avatars.githubusercontent.com/u/42048915?s=200&v=4
:width: 100px
:align: left
~~~

Nous vous proposons de l'ajouter pour vos développements web avec *Django*, *Flask* ou *FastAPI* par exemple.


*Deno* est une implémentation de JavaScript/TypeScript moderne, rapide et sécurisée.
Il permet notamment:

* Un support natif de TypeScript
* Une meilleure sécurité en gérant des permissions de manière granulaire
* Il inclut une bibliothèque standard, ainsi qu'un outil d'analyse statique et de formatage natif, un lanceur de tests, et bien plus encore
* Il est compatible avec Node et npm
* Il est écrit en langage rust et garantit ainsi rapidité et fiabilité

Et bien sûr, c'est un logiciel libre.



## Etapes d'installation

1. Installer Pixi: https://pixi.prefix.dev/latest/pixi/
2. Installer Python (via pixi)
3. Installer Jupyter et les plugins pour visualiser le cours (via pixi)
4. Installer Deno et son plugin Jupyter
5. Démarrer Jupyter Lab

Commencez par sélectionner le disque dur dans lequel vous souhaitez installer l'environnement de travail Jupyter.

En général c'est `C:`. Mais si vous avez un second disque vous pourriez vouloir utiliser ce dernier, comme `D:`

### Création de l'environnement Python

Une fois `pixi` installé, il convient de créer un environnement virtuel Python (une version de python dédiée à ce cours) puis d'y installer le serveur jupyter.

Sans quitter votre terminal, saisissez la commande suivante dans le dossier de votre choix:

~~~bash
$ pixi init formation --format pyproject
~~~

Ceci créera le dossier `formation` à l'endroit ou vous exécutez la commande.  
C'est dans ce dossier que votre environnement Python sera installé.


Positionnez-vous ensuite dans ce dossier et installez la suite Python requise

~~~
$ cd formation
$ pixi shell
(formation) $ pixi add python=3.14 jupyterlab nodejs jupyterlab-myst myst-nb
~~~

Voici la sortie attendue:
~~~
✔ Added python=3.14
✔ Added jupyterlab >=4.5.4,<5
✔ Added nodejs >=25.6.1,<25.7
✔ Added jupyterlab-myst >=2.6.0,<3
✔ Added myst-nb >=1.3.0,<2
~~~


~~~bash
$ cd formation
~~~

Activez le shell pixi pour cet environnement, qui ajoutera Python et jupyter au PATH du terminal:

~~~
$ pixi shell
(formation)  $ 
~~~

Le prompt du shell sera préfixé de `(formation) $` à partir de maintenant.

Vérifiez que la commande jupyter est bien disponible et celle de l'environnement pixi courant:
~~~
(formation) $ which jupyter
~~~

Ce qui devrait produire quelque chose comme cela:

~~~
/home/<votre login>/pixi/formation/.pixi/envs/default/bin/jupyter
~~~



## Installation de Deno

Toujours depuis votre terminal préféré, saisissez la commande suivante:

~~~bash
$ curl -fsSL https://deno.land/install.sh | sh
~~~

Nous vous invitons à accepter l'ajout de `deno` au PATH système.

La sortie devrait ressembler à ceci:

~~~
######################################################################## 100.0%   
Archive:  /home/<votre login>/.deno/bin/deno.zip
  inflating: /home/<votre login>/.deno/bin/deno  
Deno was installed successfully to /home/<votre login>/.deno/bin/deno

Deno was added to the PATH.
You may need to restart your shell for it to become available.

Run 'deno --help' to get started

Stuck? Join our Discord https://discord.gg/deno
~~~

Nous allons maintenant installer le kernel jupyter pour deno.

Pour cela, dans votre terminal, positionnez-vous dans le dossier pixi `formation`.



~~~bash
$ cd formation
~~~

Activez le shell pixi pour cet environnement, qui ajoutera Python et jupyter au PATH du terminal:

~~~
$ pixi shell
~~~

Le prompt du shell sera préfixé de `(formation) $` à partir de maintenant.

Vérifiez que la commande jupyter est bien disponible et celle de l'environnement pixi courant:
~~~
(formation) $ which jupyter
~~~

Ce qui devrait produire quelque chose comme cela:

~~~
/home/<votre login>/pixi/formation/.pixi/envs/default/bin/jupyter
~~~

Puis installez le kernel `deno`:

~~~bash
(formation)$ deno jupyter --install
Warning "deno jupyter" is unstable and might change in the future.
✅ Deno kernelspec installed successfully at /home/<votre login>/.local/share/jupyter/kernels/deno.
~~~

## Ajouter l'environnement au menu Jupyter
Enfin, afin de disposer de ce nouvel environnement dans les menus de Jupyter/Jupyterlab, il convient de l'y ajouter.

**Depuis l'environnement virtuel activé**, saisir ces commandes:


```bash
(formation) python -m ipykernel install --user --name=<nom environnement>  --display-name="<nom affiché dans Jupyter>"`
```

`<nom environnement>` représente le dossier qui sera créé dans la configuration de `ipykernel`, le nom de votre environnement sera un bon choix.

Dans notre cas, la commande finale sera:

```bash
(formation)  python -m ipykernel install --user --name=formation --display-name="Formation (Python 3.14)"
```

Redémmarez jupyter. Le nouvel environnement apparaîtra dans le menu du Launcher.
